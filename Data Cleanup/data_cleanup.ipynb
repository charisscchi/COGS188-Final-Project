## Description

Both datasets show transportation and traffic data from 8/1/2024

```python
import numpy as np
import pandas as pd

one_df = pd.read_csv("DOT_Traffic_Speeds_20261116.csv")
two_df = pd.read_csv("2024_High_Volume_FHV_Trip_Data_20261121.csv")
zones = pd.read_csv("NYC_Taxi_Zones_20261115.csv") # reference df

dot_df = one_df.copy()
fhv_df = two_df.copy() 
```

```python
# Dropping NaN rows
dot_df.dropna(inplace=True)
print(one_df.shape)
print(dot_df.shape)
fhv_df.dropna(inplace=True)
print(two_df.shape)
print(fhv_df.shape)
```

```python
# Necessary variables
new_dot_df = dot_df.loc[:, ["LINK_ID", "LINK_POINTS", "SPEED", "TRAVEL_TIME", "DATA_AS_OF", "BOROUGH"]].copy()
new_fhv_df = fhv_df.loc[:, ["request_datetime", "pickup_datetime", "dropoff_datetime", "PULocationID", "DOLocationID", "trip_miles", "trip_time"]].copy()
```

```python
# Datetime into date & time handling for FHV df

#request_datetime
new_fhv_df["RQ_date"] = new_fhv_df["request_datetime"].str[0:10]
new_fhv_df["RQ_time"] = new_fhv_df["request_datetime"].str[11:]

#pickup_datetime
new_fhv_df["PU_date"] = new_fhv_df["pickup_datetime"].str[0:10]
new_fhv_df["PU_time"] = new_fhv_df["pickup_datetime"].str[11:]

#dropoff_datetime
new_fhv_df["DO_date"] = new_fhv_df["dropoff_datetime"].str[0:10]
new_fhv_df["DO_time"] = new_fhv_df["dropoff_datetime"].str[11:]

new_fhv_df.drop(columns=["request_datetime", "pickup_datetime", "dropoff_datetime"], inplace=True)
```

```python
# DATA_AS_OF into date & time handling for DOT df

#request_datetime
new_dot_df["DATA_DATE"] = new_dot_df["DATA_AS_OF"].str[0:11]
new_dot_df["DATA_TIME"] = new_dot_df["DATA_AS_OF"].str[12:]

new_dot_df.drop(columns="DATA_AS_OF", inplace=True)
```

```python
# Giving FHV df zones a borough like in DOT df
borough = zones.set_index("Location ID")["Borough"].to_dict()

# Borough where pickup was made
new_fhv_df["PUBorough"] = new_fhv_df["PULocationID"].map(borough)

# Borough where dropoff was made
new_fhv_df["DOBorough"] = new_fhv_df["DOLocationID"].map(borough)
```
