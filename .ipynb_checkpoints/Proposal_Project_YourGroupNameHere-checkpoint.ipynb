# Abstract 
Our project aims to investigate whether traffic-aware route planning can reduce passenger pickup waiting times in a simulated robotaxi system inspired by Tesla's Cybercab. We plan to use New York City taxi trip records containing trip times and pickup and drop-off locations, along with traffic datasets containing recorded vehicle speeds and traffic volumes at different locations and times. Using OpenStreetMap to construct the road network and SUMO to simulate traffic conditions, we will compare three approaches: static A*, traffic-aware A*, and an ε-greedy learning method. We will use the data to create realistic traffic scenarios and examine how different routing strategies affect the time it takes for an assigned taxi to reach a passenger. We will evaluate their performance using simulated passenger pickup waiting time, travel time, computation time, and trip completion rate. Our goal is to determine whether incorporating traffic information and learning from previous simulated trips can improve routing efficiency compared with traditional static route planning.

Beginning in the early 2000s and 2010s, a new form of rideshare launched which allowed for drivers to provide transportation using personal vehicles as well as connecting to passengers through an app similar to a carpool. This led to better accessibility for both drivers looking to make extra money and passengers in need of easy transportation. This transformed the way people got around in large cities where “By 2015, Uber had more than 162,000 drivers, and had operations in many US cities as well as sixty countries around the world…” while competing platform “Lyft had over 60,000 drivers throughout the United States.” <a name="cit1"></a>[<sup>[1]</sup>](#cit1) One such solution that independent drivers have tackled is the wide availability of transportation anywhere one may be. Through these platforms, drivers are able to find and pick up passengers anywhere accessible by a car, as opposed to typical forms of public transportation with specific systems, stops, and wait times. As accessible as ride sharing may be, however, these drivers are still subject to obstacles on the road such as traffic congestion and relative location to passengers in need, especially in bigger cities. It was found that “In New York City, there are approximately 68,000 Uber and Lyft cars, about five times as many as yellow taxis, which has caused average speeds during business hours in Manhattan’s core to drop to a crawl...”<a name="cit2"></a>[<sup>[2]</sup>](#cit2) As a result, there is still a ways to go towards making ridesharing more efficient. One finding suggests improving distributions of drivers throughout the city so that each area may have a number of drivers within a short distance of a passenger. Through a guide, a repositioning heatmap “...highlight the driver-specific earning opportunities, approximated based on their travel time to areas with a shortage of drivers to satisfy the expected demand in the near future.”<a name="cit3"></a>[<sup>[3]</sup>](#cit3) Through this guide for drivers, the heatmap aims to direct drivers in a more efficient distribution so that they may have access to more pickup and earning opportunities. In turn, this also improves the chances that a passenger is within easy access to a driver, thus reducing wait times. 

# Problem Statement

One challenge for autonomous taxi services is getting to passengers quickly, especially when traffic conditions change throughout the day. A route that looks like the shortest or fastest option may take longer during rush hour or unexpected congestion. If a taxi follows a route without considering current traffic conditions, passengers may end up waiting longer than necessary.

Our project aims to investigate whether traffic-aware route planning can reduce passenger pickup waiting times compared with traditional static A* search. We will focus on the route an assigned taxi takes from its starting location to the passenger's pickup location, rather than changing which taxi gets assigned. Using a simulated New York City road network, we plan to compare three approaches: static A*, traffic-aware A*, and an ε-greedy learning method that selects among candidate routes based on previous simulated trips.
We will represent the road network as a graph, where intersections are nodes and road segments are edges with estimated travel-time costs. We will use real-world NYC taxi trip and traffic data to help create realistic pickup scenarios and traffic conditions. For each simulated passenger request, pickup waiting time will be calculated as the difference between the passenger's request time and the taxi's arrival time at the pickup location. Our main goal is to minimize the average pickup waiting time across multiple requests.
To compare the approaches fairly, we will test them using the same passenger requests, assigned taxis, and traffic scenarios in SUMO. We will measure average pickup waiting time, travel time, route computation time, and pickup completion rate. We will also test different values of ε to examine how exploration affects the learning-based method. By repeating the experiments under controlled conditions and recording the simulation settings and random seeds, we can compare the results and evaluate whether traffic-aware routing improves pickup efficiency.

# Proposed Solution

We propose a route-planning system for a simulated autonomous taxi traveling between pickup and drop-off areas in New York City. We will represent the road network as a graph, where intersections are nodes and road segments are edges. We plan to use NYC taxi trip records to select origin–destination areas, OpenStreetMap to construct the road network, and NYC traffic speed data to estimate travel times on monitored road segments.
We will compare the following methods:

__Static A* baseline:__ We will use A* search to find a route based on estimated travel times under uncongested conditions. This route will be selected before the trip and will not use current traffic information.

__Traffic-aware route planning:__ We will use A* with road costs based on available traffic speed or travel-time information. For road segments without matching traffic data, we will use a consistent fallback estimate. This method may avoid slower monitored roads because it considers traffic conditions when assigning costs to routes.

__ε-greedy route selection:__ For a limited set of origin–destination pairs, we will generate several candidate routes. Each candidate route will be treated as an action. After each simulated trip, the algorithm will receive a reward based on the trip’s travel time and waiting time. Routes with shorter travel and waiting times will receive higher rewards. The algorithm will update its estimated value for the selected route and use an ε-greedy policy: it will usually choose the route with the highest estimated value, while sometimes exploring another route. We will test several values of ε to examine the effect of exploration.

The bandit method may work because repeated simulated trips provide feedback about how well each candidate route performs under similar conditions. However, its estimates may not generalize to unfamiliar origin–destination pairs or traffic conditions, so we will initially evaluate it on a limited set of repeated trip scenarios.
We plan to use SUMO to simulate the trips and collect travel-time and waiting-time outputs. We will test the methods on the same selected trips and background traffic scenarios, with separate simulation runs for parameter selection and final evaluation. We will compare average travel time, waiting time, computation time, and trip completion rate. If suitable acceleration data can be generated, we will also examine ride smoothness as an additional metric.
To make the experiment reproducible, we will record the selected geographic area, data period, traffic scenarios, candidate routes, ε values, random seeds, and software versions. Before the full experiment, we will test whether the NYC traffic data can be matched to the road network. If the match is insufficient, we will use simulated traffic scenarios and clearly report that result.
