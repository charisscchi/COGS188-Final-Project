# Background

Ride-sharing services such as Uber and Lyft have changed how people travel in large cities by allowing passengers to request transportation through mobile applications. These services have made transportation more accessible, but vehicles still face challenges such as traffic congestion, travel delays, and the distance between drivers and passengers [1]. In New York City, high levels of ride-sharing activity and congestion can make it difficult for vehicles to reach their destinations efficiently [2]. These problems are especially important for robotaxi services because passengers depend on the system to select routes that allow assigned vehicles to reach pickup locations quickly.

Previous research has explored different ways to improve ride-sharing efficiency. Haferkamp, Ulmer, and Ehmke (2024) studied heatmap-based decision support for repositioning drivers toward areas with expected passenger demand [3]. Their work shows how routing and vehicle-location decisions can affect ride-sharing operations. However, our project focuses on a different part of the problem: after a taxi has already been assigned to a passenger, how should it choose the route to reach the pickup location?

One approach to route planning is A*, a heuristic search algorithm that finds minimum-cost paths in a graph. Hart, Nilsson, and Raphael (1968) established the theoretical foundation of A* search [4]. Traditional A* can find an efficient route using fixed road costs, but these costs may not reflect changing traffic conditions. For example, a road that normally provides the shortest travel time may become slower during rush hour.

Our project investigates whether incorporating traffic information into route planning can help reduce passenger pickup waiting times. We plan to compare static A* with traffic-aware A* and an ε-greedy learning method that selects among candidate routes based on feedback from repeated simulated pickup attempts. Using a New York City road network and simulated traffic conditions, we aim to evaluate how traffic information and learning affect routing efficiency.

# Problem Statement

One challenge for autonomous taxi services is reaching passengers quickly, especially when traffic conditions change throughout the day. A route that appears shortest or fastest under normal conditions may take longer during rush hour or unexpected congestion. If a taxi follows a route without considering traffic conditions, passengers may end up waiting longer than necessary.

Our project investigates whether traffic-aware route planning can reduce passenger pickup waiting times compared with traditional static A* search. We focus on the route an already assigned taxi takes from its starting location to the passenger's pickup location, rather than changing which taxi is assigned. We will compare three approaches: static A*, traffic-aware A*, and an ε-greedy learning method that selects among candidate routes based on feedback from previous simulated pickup attempts.

We represent the New York City road network as a graph, where intersections are nodes and road segments are edges with estimated travel-time costs. We will use historical taxi trip records and traffic measurements to help construct realistic pickup scenarios and traffic conditions. Since the historical data does not provide the locations of all available taxis, we will generate taxi starting positions using a consistent simulation procedure.

For each simulated passenger request, pickup waiting time will be calculated as the difference between the passenger's request time and the assigned taxi's arrival time at the pickup location:

$$
W_i = t_i^{\text{arrival}} - t_i^{\text{request}}
$$

Our main objective is to minimize average passenger pickup waiting time across multiple simulated requests. To compare the approaches fairly, we will use the same passenger requests, assigned taxi starting positions, and traffic scenarios in SUMO. We will also measure route travel time, traffic waiting time, computation time, and pickup completion rate.

We will repeat experiments under controlled conditions, record simulation settings and random seeds, and evaluate different values of ε to examine how exploration affects the learning-based method. This will allow us to compare the routing methods using measurable and reproducible results.

# Proposed Solution

Our project will develop and compare three route-planning approaches for a simulated robotaxi system: static A*, traffic-aware A*, and ε-greedy route selection. The goal is to determine whether incorporating traffic information and learning from previous simulated pickup attempts can reduce passenger pickup waiting times compared with traditional static routing.

### 1. Road Network Representation

We will use OpenStreetMap to construct a road network for a selected area of New York City. The network will be represented as a graph, where intersections are nodes and road segments are edges.

Each edge will have an estimated travel-time cost. For the static A* baseline, these costs will remain fixed. For traffic-aware A*, we will use available traffic-speed information to adjust the estimated costs of road segments.

We will use NYC High Volume FHV trip records to generate passenger pickup scenarios and simulate the starting locations of available taxis. Each request will have a fixed assigned taxi, starting position, and pickup location so that the routing methods can be compared under the same conditions.

### 2. Static A* Baseline

Our baseline will use A* search to find a minimum-cost route from the assigned taxi's starting position to the passenger's pickup location.

We will calculate edge costs using fixed estimated travel times under uncongested conditions. The algorithm will use these costs and a heuristic estimate of the remaining travel time to select a route.

Once a route is selected, the baseline will not adjust its road costs using current traffic information. This gives us a reference for evaluating whether traffic-aware routing provides an improvement.

### 3. Traffic-Aware A*

The second approach will also use A* search, but its edge costs will incorporate available traffic-speed or travel-time measurements.

Road segments with lower measured speeds may receive higher estimated travel-time costs. This allows the algorithm to consider alternative routes that may be longer in distance but faster under the current traffic conditions.

For road segments without matching traffic measurements, we will use a consistent fallback travel-time estimate. We will document how traffic observations are matched to the road network and how missing measurements are handled.

We will compare the routes selected by traffic-aware A* with those selected by static A* under the same simulated traffic scenarios.

### 4. ε-Greedy Route Selection

Our third approach will use an ε-greedy learning method to select among several candidate routes between a simulated taxi starting position and a passenger pickup location.

For a limited set of repeated pickup scenarios, we will generate multiple candidate routes. Each candidate route will be treated as an action, and the algorithm will maintain an estimated value for each route based on previous simulated outcomes.

After each pickup attempt, the algorithm will receive a reward based on the time required for the taxi to reach the passenger. Routes producing shorter pickup waiting times will receive higher rewards.

The ε-greedy policy will balance exploration and exploitation:

- **Exploitation:** With probability $1-\varepsilon$, the algorithm selects a route with the highest estimated reward.
- **Exploration:** With probability $\varepsilon$, it selects a route randomly to gather additional information.

The estimated value of a selected route will be updated using the rewards observed from repeated simulations. We will test different values of $\varepsilon$ to examine how exploration affects performance.

Because this method learns from repeated experiences, we will initially evaluate it on a limited set of recurring pickup scenarios. Its performance may not generalize to unfamiliar routes or traffic conditions.

# Evaluation Metrics

We will evaluate static A*, traffic-aware A*, and ε-greedy route selection using the same passenger requests, assigned taxi starting locations, and simulated traffic conditions. Our primary metric will be average passenger pickup waiting time because our main goal is to determine whether traffic-aware routing helps taxis reach passengers faster.

We will also measure route travel time, traffic waiting time, computation time, and pickup completion rate. These metrics will allow us to compare routing efficiency, computational cost, and reliability.

### 1. Average Passenger Pickup Waiting Time

Passenger pickup waiting time measures how long a passenger waits between requesting a taxi and the assigned taxi arriving at the pickup location.

For each completed pickup, we calculate:

$$
W_i^{\text{pickup}} = t_i^{\text{arrival}} - t_i^{\text{request}}
$$

The average across $N$ completed pickups is:

$$
\overline{W}_{\text{pickup}} =
\frac{1}{N}\sum_{i=1}^{N} W_i^{\text{pickup}}
$$

where $t_i^{\text{request}}$ is when the passenger requests a taxi, and $t_i^{\text{arrival}}$ is when the assigned taxi reaches the pickup location.

A lower average means passengers spend less time waiting. This will be our main metric for comparing the three routing approaches.

### 2. Average Route Travel Time

Route travel time measures how long the assigned taxi takes to travel from its starting location to the passenger's pickup location.

For each completed pickup:

$$
T_i^{\text{route}} =
t_i^{\text{arrival}} - t_i^{\text{departure}}
$$

The average route travel time is:

$$
\overline{T}_{\text{route}} =
\frac{1}{N}\sum_{i=1}^{N} T_i^{\text{route}}
$$

where $t_i^{\text{departure}}$ is the taxi's departure time.

A lower average indicates that the taxi reaches the pickup location more quickly after departure.

If taxis depart immediately when requests are made, route travel time and passenger pickup waiting time will be equal. We will document this assumption when reporting our results.

### 3. Average Traffic Waiting Time

Traffic waiting time measures how much time a taxi spends nearly stopped during its route. We plan to use SUMO's waiting-time output, which counts time when a vehicle's speed is at or below 0.1 m/s, excluding scheduled stops.

The average is:

$$
\overline{W}_{\text{traffic}} =
\frac{1}{N}\sum_{i=1}^{N} W_i^{\text{traffic}}
$$

where $W_i^{\text{traffic}}$ is the traffic waiting time for completed pickup attempt $i$.

A lower average indicates less time spent nearly stopped. This metric is different from passenger pickup waiting time and helps us understand how congestion affects the selected routes.

### 4. Average Computation Time

Computation time measures how long an algorithm takes to select a route.

We will calculate:

$$
\overline{C} =
\frac{1}{M}\sum_{i=1}^{M} C_i
$$

where $C_i$ is the route-selection time for attempt $i$, and $M$ is the number of route-planning attempts.

We will use the same timing procedure for each method. For the ε-greedy approach, training time will be reported separately from route-selection time.

This metric will help us examine whether improvements in pickup efficiency require additional computational cost.

### 5. Pickup Completion Rate

Pickup completion rate measures the percentage of attempted pickups in which the assigned taxi successfully reaches the passenger's pickup location within the simulation time limit.

$$
\text{Pickup Completion Rate} =
\frac{\text{Number of completed pickups}}
{\text{Number of attempted pickups}}
\times 100\%
$$

A higher completion rate indicates that more pickup requests were successfully completed.

This metric is important because an algorithm should not appear better simply because difficult or unfinished pickup attempts are excluded from average waiting-time calculations.

### 6. Model Selection and Additional Evaluation

For the ε-greedy method, we will test several exploration rates to examine how the choice of $\varepsilon$ affects passenger pickup waiting time and completion rate.

We will use separate simulation runs for parameter selection and final evaluation so that we do not evaluate the learning method only on the scenarios used to select its settings.

If feasible, we will also examine simulated acceleration and braking as an additional measure of route smoothness. However, these measurements will not be treated as direct evidence of passenger comfort without actual passenger feedback.

### 2. Limitations

Our project has several potential limitations related to the available datasets, simulation environment, and routing methods.

**Traffic Data Coverage:** NYC DOT traffic speed measurements are only available for monitored road segments. Some roads in our selected simulation area may not have matching traffic observations. Using fallback travel-time estimates could affect the accuracy of traffic-aware routing, especially in areas with limited coverage.

**Simulated Taxi Locations:** The High Volume FHV trip records contain passenger request times and pickup areas but do not provide the locations of all available taxis. Because of this, we will generate taxi starting positions in the simulation. These positions may not fully represent how real robotaxi fleets are distributed throughout the city.

**Geographic Accuracy:** Historical pickup locations are represented by taxi zone IDs rather than exact GPS coordinates. We will use representative pickup locations generated from the taxi zones and road network, which may not perfectly match real pickup positions.

**Simulation Limitations:** SUMO provides a controlled environment for comparing routing algorithms, but simulated traffic conditions may differ from real-world traffic patterns. Our experiments will not capture every factor that affects an actual autonomous taxi, such as unexpected road closures, human driving behavior, or real-time safety decisions.

**Learning-Based Routing:** Our ε-greedy approach will initially be evaluated on a limited set of repeated pickup scenarios. Its learned route preferences may not generalize well to unfamiliar locations or rapidly changing traffic conditions.

These limitations will be revisited after implementation to distinguish problems we actually encountered from potential limitations of the experimental design.

### 3. Future Work

There are several possible directions for improving our robotaxi routing system beyond the initial project.

**More Complete Traffic Information:** Future work could incorporate additional real-time or historical traffic measurements to improve road travel-time estimates and reduce reliance on fallback values.

**Dynamic Route Updates:** Our initial comparison focuses on selecting routes using available traffic information. A future version could allow taxis to update their routes during travel when traffic conditions change unexpectedly.

**Improved Learning Methods:** We could investigate contextual bandits or reinforcement learning methods that incorporate information such as time of day, location, and current traffic conditions. These approaches may help the system adapt to a wider range of pickup scenarios.

**Larger Simulation Areas:** Expanding the simulation to more neighborhoods and different traffic conditions could help evaluate whether the routing methods perform consistently across New York City.

**Taxi Assignment and Fleet Management:** Our current project keeps taxi assignments fixed to isolate the effect of route planning. Future work could also investigate how assigning passengers to different available taxis affects pickup waiting times.

The most useful next steps will depend on which limitations and performance differences we observe during the experiments.

We will discuss these concerns when interpreting our findings and recognize that optimizing pickup efficiency alone is not sufficient for evaluating a real-world autonomous transportation system.
